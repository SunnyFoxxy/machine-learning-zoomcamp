Q7. Sum of weights

```python
import numpy as np
X = df.loc[df['origin'] == 'Asia', ['vehicle_weight', 'model_year']].head(7).values 
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv.dot(X.T).dot(y)
w.sum()
```
